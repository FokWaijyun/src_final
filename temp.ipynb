```python
import numpy as np

imp={ 1:1, 2:2, 3:3 }
kg={
    1:[(0,0),(1,1),(2,3)],
    2:[(3,3),(4,4)],
    3:[(5,5)]
    }

t2= np.array([1,1,1,2,2,3])
t1= np.array([imp[i] for i in t2]) # 头节点对应的重要性
t3= np.array([i for i in range(6)])
t4= np.array([i for i in range(6)])
l = zip(t1,t2 ,t3 ,t4 )
l = sorted(l)
# 'unzip'
# try:
index=np.argsort(-np.array(t1))
r2= [t2[i] for i in index]
r3= [t3[i] for i in index]
r4= [t4[i] for i in index]

# 采样前n个不同节点
need=3
count=0
nodes=[]
for i in r2:
    if count>=need:
        break
    if i not in nodes:
        nodes.append(i)
        count+=1

# 计算前n个节点重要性比例
import math
nodes_imp=[imp[i] for i in nodes]
total=sum(nodes_imp)
nodes_imp_rate=[i/total*need for i in nodes_imp]
nodes_imp_rate_approximate=[math.ceil(i/total*need) for i in nodes_imp]
##如果超过，逆序逐个减1
remain=sum(nodes_imp_rate_approximate)-need
while remain:
    if nodes_imp_rate_approximate[-(i+1)] >1:
        nodes_imp_rate_approximate[-(i+1)]-=1
    else:
        continue

node2num={}
for index,head in enumerate(nodes):
    if nodes_imp_rate_approximate[index]<=0:
        break
    else:
        node2num[head]=nodes_imp_rate_approximate[index]

# 获取头节点对应的知识集，随机抽取对应个数知识。
## 少于比例则重复抽取
h=[]
r=[]
t=[]
for head,num in node2num.items():
    replace = len(kg[head]) < num # 如果超过采样数量则使用重点采样
    head_knowledge_set=kg[head]
    indices = np.random.choice(len(kg[head]), size=num, replace=replace)
    for i in indices:
        h.append(head)
        r.append(head_knowledge_set[i][1])
        t.append(head_knowledge_set[i][0])
    # break






print(np.argsort(-t1))
# num=3
# print(r2[0:num])
# print(r3[0:num])
# print(r4[0:num])
```
